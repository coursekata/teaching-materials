## <center>Quick Reference Math Notation</center>


NOTE: To indicate a mathematical expression in markdown, surround the expression in dollar signs (\$). 

- If you surround it in two dollar signs, it will drop the expression to the next line and render the expression in the center of the cell, for example: 

$$Y_i = b_0 + e_i$$ 

- If you only use one dollar sign on each side, it will not center it and will keep it on the same line of text, as so: $Y_i = b_0 + e_i$.

---

### GLM Notation:

> Empty Model (sample notation):

$$Y_i = b_0 + e_i$$

> Empty Model: (parameter notation): 

$$Y_{i} = \beta_{0} + \epsilon_{i}$$

> Complex Model (2 parameters)

$$Y_i = b_0 + b_1X_{i} + e_i$$

> Complex Model (3 parameters)

$$Y_i = b_0 + b_1X_{1i} + b_2X_{2i} + e_i$$

> Complex Model (4 parameters)

$$Y_i = b_0 + b_1X_{1i} + b_2X_{2i} + b_3X_{3i} + e_i$$

### Mathematical Formulas:

> Sum of residuals (from the mean):

$$\sum_{i=1}^n (Y_i - \bar{Y})$$

> Sum of squares (from the mean):

$$SS = \sum (Y_i - \bar{Y})^2$$

> Variance:

$$\sigma^2 = \frac {\sum (Y_i - \bar{Y})^2}{n-1}$$

> Standard Deviation:

$$\sigma = \sqrt \frac {\sum (Y_i - \bar{Y})^2}{n-1}$$

> Standard Error:

$$\sigma_{\bar{Y}} = \frac {\sigma}{\sqrt{n}}$$

> PRE (Proportinal Reduction in Error):

$$PRE = \frac {SS_{\text{Model}}}{SS_{\text{Total}}}$$

> F Ratio/Variance Ratio:

$$F = \frac {MS_{\text{Model}}}{MS_{\text{Error}}}$$

## <center>General Text Formatting</center>

#### <center>Code in Markdown Cells:</center>

If you want to show students some code in a markdown cell, use the tick mark (which is usually with the ~ key on your keyboard). 

Three tick marks creates a little section of code:
```
gf_histogram(~Thumb, data=Fingers)%>%
gf_facet_grid(Sex~.)
```

If you want to show that something is code in a sentence, just surround the code with one tick mark like this: `gf_histogram()` is a function we use.


#### <center>Linking Text:</center>

Insert a hyperlink in your text with brackets and parentheses, like this:

[link text](http://url)
