# How to plot residuals and squared residuals

This is the long version of the code that can be used to plot the residuals and squared residuals of a few specific data points on a graph for pedagogical purposes.

However, you can now use the functions `gf_resid()` and `gf_square_resid()` as a short cut (they are automatically included when you run `library(coursekata)`), however, these functions will plot every data point (rather than specified data points).

```r
library(coursekata)
```

```r
# image res
options(repr.plot.res=300)

# new colors
blue80 <- "#002d9c"

ck_alpha_bar <- .7
ck_alpha_point <- .6
```

```r
Fingers$TinySet <- c(1,1,1,0,0,0,1,0,0,1, rep(0,147))
Fingers$TinySet[142] <- 1

empty_model <- lm(Thumb ~ NULL, data = Fingers)
Fingers$Predict <- predict(empty_model)
Fingers$Resid <- resid(empty_model)
```

```r
# Generate Residuals
Fingers$prediction_empty <- predict(empty_model)
Fingers$residual_empty <- resid(empty_model)

tiny_df <- filter(Fingers, TinySet == 1)

g1_empty <- gf_point(Thumb ~ Height, data = Fingers, width = .2, alpha = ~TinySet, show.legend = FALSE) %>%
  gf_model(empty_model, color = blue80) + 
  scale_alpha(range = c(0.08, 1))

s_color <- blue80
s <-.5


scale <- 1/4.75
tiny_df$sq.x_side2_empty <- tiny_df$Height + abs(tiny_df$residual_empty*scale)

g1_empty +
  geom_segment(aes(x = tiny_df$Height[1], y = tiny_df$prediction_empty[1], xend = tiny_df$Height[1], yend = tiny_df$Thumb[1]), color = s_color, size = s, show.legend = FALSE) +  
  geom_segment(aes(x = tiny_df$Height[2], y = tiny_df$prediction_empty[2], xend = tiny_df$Height[2], yend = tiny_df$Thumb[2]), color = s_color, size = s, show.legend = FALSE) +  
  geom_segment(aes(x = tiny_df$Height[3], y = tiny_df$prediction_empty[3], xend = tiny_df$Height[3], yend = tiny_df$Thumb[3]), color = s_color, size = s, show.legend = FALSE) +  
  geom_segment(aes(x = tiny_df$Height[4], y = tiny_df$prediction_empty[4], xend = tiny_df$Height[4], yend = tiny_df$Thumb[4]), color = s_color, size = s, show.legend = FALSE) +  
  geom_segment(aes(x = tiny_df$Height[5], y = tiny_df$prediction_empty[5], xend = tiny_df$Height[5], yend = tiny_df$Thumb[5]), color = s_color, size = s, show.legend = FALSE) +  
  geom_segment(aes(x = tiny_df$Height[6], y = tiny_df$prediction_empty[6], xend = tiny_df$Height[6], yend = tiny_df$Thumb[6]), color = s_color, size = s, show.legend = FALSE) 
```
